## 1. Mathematical Formulation & Extensive-Form Game Dynamics

The Kaggle **Kaggriculture** environment is formalized as a finite-horizon, two-player simultaneous extensive-form game with imperfect information and stochastic weather transitions:

$$\Gamma = \langle \mathcal{N}, \mathcal{S}, \{\mathcal{A}_i\}_{i \in \mathcal{N}}, \mathcal{T}, \{\mathcal{R}_i\}_{i \in \mathcal{N}}, \{\Omega_i\}_{i \in \mathcal{N}}, \mathcal{O}, T \rangle$$

Where:
- $\mathcal{N} = \{0, 1\}$ denotes the set of symmetric competing agents.
- $\mathcal{S}$ is the underlying Markovian state space encompassing spatial grid layouts, market order books, capital allocations, inventory counts, and weather forecasts.
- $T = 720$ defines the discrete time horizon, structured into $D = 30$ diurnal cycles with $\tau = 24$ hourly ticks per day:

$$t = 24 \cdot d + h, \quad d \in \{0, \dots, 29\}, \; h \in \{0, \dots, 23\}$$

### 1.1 Spatial Geometry & Territory Expansion
competing players operate on disjoint, symmetric territorial domains $\mathcal{G}^{(p)} \subset \mathbb{Z}^{10 \times 10}$. The farm domain initializes with the central quadrant active:

$$\mathcal{Q}_0 = \{ (x, y) \in \mathbb{Z}^2 \mid 2 \le x \le 7, \; 2 \le y \le 7 \}$$

Territorial expansion occurs through discrete quadrant acquisitions $\mathcal{Q}_k \subset \mathcal{G}^{(p)}$ subject to the monotonically convex acquisition cost schedule:

$$\mathcal{C}_{\text{land}}(k) = 1000 \cdot 2^{k-1}, \quad k \in \{1, 2, 3\}$$

Unlocking quadrant $k$ expands available arable plot surface area:

$$|\mathcal{A}_{\text{farm}}(k)| = 25 \cdot (k + 1) \quad \implies \quad \max |\mathcal{A}_{\text{farm}}| = 100 \text{ tiles}$$

### 1.2 Multi-Agent Action Tuple Formulation
At each discrete step $t \in [0, T-1]$, each player $p \in \mathcal{N}$ simultaneously executes a compound action vector:

$$\mathbf{a}_t^{(p)} = \left\langle \mathbf{u}_{\text{farmer}}^{(p)}, \; \{\mathbf{u}_{\text{hand}, i}^{(p)}\}_{i=1}^{H_t^{(p)}}, \; \mathbf{m}_t^{(p)} \right\rangle \in \mathcal{A}^{(p)}$$

Where:
- $\mathbf{u}_{\text{farmer}}^{(p)} \in \mathcal{U}$ defines the spatial operator assigned to the primary farmer unit.
- $\mathbf{u}_{\text{hand}, i}^{(p)} \in \mathcal{U}$ specifies the independent operator for each hired laborer $i \in \{1, \dots, H_t^{(p)}\}$.
- $\mathcal{U} = \{\text{MOVE}(d), \text{DIG}, \text{PLANT}(c), \text{WATER}, \text{FERTILIZE}, \text{HARVEST}, \text{FEED}, \text{CARE}, \text{PASS}\}$ represents the discrete unit operator set.
- $\mathbf{m}_t^{(p)} = \left[ o_1^{(p)}, o_2^{(p)}, \dots, o_K^{(p)} \right]$ denotes the bounded market order queue ($K \le 10$).

````python
# Run the real file loader in both seats before creating a submission.
import tempfile
from pathlib import Path
from importlib.metadata import version
from kaggle_environments import make
from kaggle_environments.agent import get_last_callable

print('Engine:', version('kaggle-environments'))
print('Entry point:', get_last_callable(SOURCE.decode()).__name__)
assert get_last_callable(SOURCE.decode()).__name__ == 'step1010_step1009_fixed_point_agent'
with tempfile.TemporaryDirectory() as directory:
    path = Path(directory) / 'main.py'
    path.write_bytes(SOURCE)
    for seat in (0, 1):
        agents = ['starter', 'starter']
        agents[seat] = str(path)
        env = make('kaggriculture', configuration={'episodeSteps': 720, 'seed': 290901})
        env.run(agents)
        final = env.steps[-1]
        rewards = [state.reward for state in final]
        assert all(state.status == 'DONE' for state in final)
        assert rewards[seat] > rewards[1-seat], ('Starter sanity check failed', seat, rewards)
        print('Seat', seat, 'final cash:', rewards)
# These sanity checks do not estimate the live leaderboard rating.
````
